```python
#8
def is_palidrome(n):
    return str(n) == str(n)[::-1]
def reverse_number(n):
    return int(str(n)[::-1])
def reverse_add_palidrome(n):
    count = 0
    while not is_palidrome(n):
        n += reverse_number(n)
        count += 1
    return count, n
n = int(input())
count, result = reverse_add_palidrome(n)
print(f"Count: {count}, Result: {result}")
```

```python
#w10
def collatz(n):
    length=1
    while n!=1:
        if n%2==0:
            n=n//2
        else:
            n=n*3+1
        length += 1
    return length
def longest_collatz(n):
     max_len=0
     a=1
     for i in range(1,n+1):
         len=collatz(i)
         if len > max_len:
             max_len=len
             a=i
     print(a,max_len)
longest_collatz(int(input()))
```

```python
#w13
a, b = map(int, input().split())

sumA = 0
sumB = 0
for i in range(1, a):
    if a % i == 0:
        sumA += i
for i in range(1, b):
    if b % i == 0:
        sumB += i
if sumA == b and sumB == a:
    print("true")
else:
    print("false")
```
