```python
#21
words = input().split()

pos_dict = {}

for i, word in enumerate(words):
    if word not in pos_dict:
        pos_dict[word] = []
    pos_dict[word].append(i)

for word in pos_dict:
    pos_dict[word] = tuple(pos_dict[word])

print(pos_dict)
```
