```python
#1
def binary_search(arr,value):
    left=0
    right=len(arr)-1
    while left<= right:
        mid=(left+right)//2
        if arr[mid] == value:
            return mid
        elif arr[mid]<value:
            left=mid+1
        else:
            right=mid-1
    return -1
arr=list(map(int,input().split()))
value=int(input())
print(binary_search(arr,value))
```

```python
#3
def bubble_sort(lst):
    a=len(lst)
    for i in range(a):
        b=False
        for j in range(a-i-1):
            if lst[j]>lst[j+1]:
                lst[j],lst[j+1]=lst[j+1],lst[j]
            b=True
        if not b:
            break
    return lst
lst=list(map(int,input().split()))
print(bubble_sort(lst))

```

```python
#10
import bisect

def max_subarray_sum_mod(nums, m):
    prefix = 0
    max_sum = 0
    seen = []

    for x in nums:
        prefix = (prefix + x) % m
        max_sum = max(max_sum, prefix)
        
        idx = bisect.bisect_right(seen, prefix)
        if idx < len(seen):
            candidate = (prefix - seen[idx] + m) % m
            max_sum = max(max_sum, candidate)

        bisect.insort(seen, prefix)

    return max_sum
```
