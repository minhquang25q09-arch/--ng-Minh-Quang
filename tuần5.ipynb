```python
#w1
def max_of_two(a,b):
    return max(a,b)
a,b=map(int,input().split())
print(max_of_two(a,b))
```

```python
#w3
def is_prime(a):
  if a>0 and all(a%i!=0 for i in range(2,int(a**0.5)+1)):
    return True
  return False
a=int(input())
print(is_prime(a))
 
```

```python
#w4
def perfect_number(a):
    tong_cac_uoc=0
    for i in range(1,a):
        if a%i==0:
            tong_cac_uoc+=i
    if tong_cac_uoc == a:
        return True
    else:
        return False
print(perfect_number(int(input())))
```

```python
#w6
def giaithua(a):
    b=1
    for i in range(1,a+1):
        b*=i
    return b
print(giaithua(int(input())))
    
```

```python
#w5
def find_first_index(lst,a):
    for i in range(0,len(lst)+1):
        if lst[i] == a:
            return i+1
        else:
            return -1
lst=list(map(int,input().split()))
a=int(input())
print(find_first_index(lst,a))
```

```python
#10
def isomorphic(a, b):
    map_ab = {}
    map_ba = {}
    for x, y in zip(a, b):
        if x in map_ab:
            if map_ab[x] != y:
                return False
        else:
            map_ab[x] = y
        if y in map_ba:
            if map_ba[y] != x:
                return False
        else:
            map_ba[y] = x

    return True
print(isomorphic(input().split()))
```
